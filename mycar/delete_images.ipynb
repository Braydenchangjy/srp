```python
import os

# Function to delete images based on index range
def delete_images(image_directory, start_index, end_index):
    """
    Deletes images in the specified range [start_index, end_index] from the given directory.
    
    Parameters:
        image_directory (str): Path to the directory containing the images.
        start_index (int): Starting index of images to delete.
        end_index (int): Ending index of images to delete.
    """
    for index in range(start_index, end_index + 1):
        image_filename = f"{index}_cam_image_array_.jpg"
        image_path = os.path.join(image_directory, image_filename)
        if os.path.exists(image_path):
            os.remove(image_path)
            print(f"Deleted: {image_path}")
        else:
            print(f"File not found: {image_path}")

# Define the directory containing the images
image_directory = './data_loop/house/2/images'  # Replace with the path to your images folder

# Specify the range of images to delete
start_index = 11975
end_index = 12092



# Perform the deletion
delete_images(image_directory, start_index, end_index)
```
